### The synthetic dataset has 10 versions with different levels of noise

````python
x_paths = sorted(
    [_ for _ in DATASETS["synthetic"].dir.glob("*.tif") if "clean" not in _.stem and "deepcad" not in _.stem],
    key=lambda _: float(_.stem.split("_")[-2].split("dBSNR")[0]),
)
cprint(*[f"rand:{' - '.join(_.stem.split('_')[1:-1])}" for _ in x_paths], sep="\n")
````

````python
y_path = DATASETS["synthetic"].dir / "clean_30Hz_490x490x6000.tif"
y = Recording(y_path, max_frames=1).normalized
xs = {f.stem[:-13]: Recording(f, max_frames=1).normalized for f in x_paths}
````

````python
imshow({**xs, "Ground Truth": y}, cols=4, vrange=(0,1), size=8)
````

In this context, I will consider the wrost case, the $Q=1$. Here the SNR is as low as in several real-life experiments, expecially those using small equipment like the mobile 2PM used by the OABF research group.

**The low input SNR affects the predictions of the SOTA denoisers**. They look somewhat blurry and Freq2Clean has good margin of improvement.

## PSNR and SSIM
My PSNR calculation is made with data normalized in [0,1] so it won't be negative (if `data_range=1`).

Both PSNR and SSIM are here computed on a single frame basis. The plot below visualizes the trend of the two metrics in the 10 versions.

````python
psnrs = {k.split("_")[1].split("_")[0]: psnr(y, x, data_range=1) for k, x in tqdm(xs.items())}
ssims = {k.split("_")[1].split("_")[0]: ssim(y, x, data_range=1) for k, x in tqdm(xs.items())}
````

````python
df = pd.DataFrame({"PSNR (dB)": psnrs, "SSIM": ssims})
fig, ax1 = plt.subplots(figsize=(12, 6))
df["PSNR (dB)"].plot(marker="o", ax=ax1, color="tab:blue", label="PSNR (dB)")
ax1.set_ylabel("PSNR (dB)")
ax1.set_ylim(0, 30)
ax1.set_yticks(range(0, 31, 6))
ax1.set_xticks(range(len(df)))

ax2 = ax1.twinx()
df["SSIM"].plot(marker="s", ax=ax2, color="tab:orange", label="SSIM")
ax2.set_ylabel("SSIM")
ax2.set_ylim(0, 1)

h1, l1 = ax1.get_legend_handles_labels()
h2, l2 = ax2.get_legend_handles_labels()
ax1.legend(h1 + h2, l1 + l2, loc="lower right")

plt.title("PSNR vs SSIM")
plt.show()
````

The $Q=1$ version has a $PSNR\approx 1dB$. How further can we push it using self-supervised video denoising?
